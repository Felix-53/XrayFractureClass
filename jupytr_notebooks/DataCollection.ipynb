```python
import os
def remove_non_image_file(my_data_dir):
    image_extension = ('.png', '.jpg', '.jpeg') # Allowed image file options
    
    folders = os.listdir(my_data_dir) # List of all in the file
    for folder in folders: #Loops over all in the folders
        folder_path = my_data_dir + '/' + folder # Path to the file/item
        if not os.path.isdir(folder_path): #Checks the items not a directory
           

             continue
    
    
    files = os.listdir(folder_path)
    i = [] # List to count non image files
    j = [] # List to count image files
    for given_file in files: # loops over all in the folders
           
            if not given_file.lower().endswith(image_extension): # Checks the file is not an img
                file_location = my_data_dir + '/' + folder + '/' + given_file
                os.remove(file_location) # Deletes the non image files
                i.append(1)
            else:
                j.append(1)
    print(f"Folder: {folder} - has image file", len(j))
    print(f"Folder: {folder} - has non-image file", len(i))
```

```python
def split_train_validation_test_images(my_data_dir, train_set_ratio, validation_set_ratio, test_set_ratio): # Splits the image into sub sections train test val

    if train_set_ratio + validation_set_ratio + test_set_ratio != 1.0:
        print("train_set_ratio + validation_set_ratio + test_set_ratio should sum to 1.0") # Needs to add to over 1 or wont work
        return
    labels = os.listdir(my_data_dir) 
   
   
    if 'test' in labels: # Skips the split if done already
        pass
    else:
        for folder in ['train', 'validation', 'test']:
            for label in labels:
                os.makedirs(name=my_data_dir + '/' + folder + '/' + label) # Makes subfolders
        for label in labels:
            files = os.listdir(my_data_dir + '/' + label)
            random.shuffle(files) # Shuffles the files to prev cnn learning patterns
            train_set_files_qty = int(len(files) * train_set_ratio) # Amount of files going to train
            validation_set_files_qty = int(len(files) * validation_set_ratio) # Amount to val and remaining to test 
            count = 1
           
           
            for file_name in files: # The loop
                if count <= train_set_files_qty:
                    shutil.move(my_data_dir + '/' + label + '/' + file_name, # Moves the file
                                my_data_dir + '/train/' + label + '/' + file_name) # Tp train
                
                elif count <= (train_set_files_qty + validation_set_files_qty):
                    shutil.move(my_data_dir + '/' + label + '/' + file_name, # Moves the file
                                my_data_dir + '/validation/' + label + '/' + file_name) # To val
                else:
                    shutil.move(my_data_dir + '/' + label + '/' + file_name, # Moves the file
                                my_data_dir + '/test/' + label + '/' + file_name) # To test
                count += 1
           
           
            os.rmdir(my_data_dir + '/' + label) # Deletes the original data file
```
